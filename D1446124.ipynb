```python
# --- 測試一：原始資料 ---
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 1. 讀取活動名稱，建立兩份整理後（去半形空白、轉小寫、保留重複）的串列
board_activity, movie_activity = activity_names
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 2. 轉成集合去重並做集合運算
board_set = set(board_list)
movie_set = set(movie_list)

both_sorted = sorted(board_set & movie_set)          # 共同參加者（交集）
all_sorted = sorted(board_set | movie_set)           # 全部參加者（聯集）
board_only_sorted = sorted(board_set - movie_set)    # 只參加桌遊者（差集）
movie_only_sorted = sorted(movie_set - board_set)    # 只參加電影者（差集）

# 3. 通知名單預覽（全部參加者排序後取前兩位）
notify_preview = all_sorted[:2]

# 4. 建立 summary 字典，計算人數與應收費用（金額保留數值）
board_count = len(board_set)
movie_count = len(movie_set)
both_count = len(board_set & movie_set)
total_unique_count = len(board_set | movie_set)
total_visits = board_count + movie_count

board_revenue = board_count * unit_fees[board_activity]
movie_revenue = movie_count * unit_fees[movie_activity]
total_revenue = board_revenue + movie_revenue

summary = {
    "board_count": board_count,
    "movie_count": movie_count,
    "both_count": both_count,
    "total_unique_count": total_unique_count,
    "total_visits": total_visits,
    "board_revenue": board_revenue,
    "movie_revenue": movie_revenue,
    "total_revenue": total_revenue,
}

# 5. 輸出完整摘要
print("HW09 活動報名摘要（測試一：原始資料）")
print(f"活動：{board_activity}，每人 {unit_fees[board_activity]:.2f} 元")
print(f"活動：{movie_activity}，每人 {unit_fees[movie_activity]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{sorted(board_set)}")
print(f"電影去重名單：{sorted(movie_set)}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{both_sorted}")
print(f"共同人數：{summary['both_count']} 人")
print(f"只參加桌遊者：{board_only_sorted}")
print(f"只參加電影者：{movie_only_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['total_unique_count']} 人")
print(f"報名人次：{summary['total_visits']} 人次")
print(f"通知名單前兩位：{notify_preview}")
print(f"桌遊應收：{summary['board_revenue']:.2f} 元")
print(f"電影應收：{summary['movie_revenue']:.2f} 元")
print(f"總應收：{summary['total_revenue']:.2f} 元")
```
